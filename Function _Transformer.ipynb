```python
sns.displot(dataset["CoapplicantIncome"])
plt.show()
```

```python
q1 = dataset["CoapplicantIncome"].quantile(0.25)
q3 = dataset["CoapplicantIncome"].quantile(0.75)
IQR = q3-q1
```

```python
min_range = q1-(1.5*IQR)
max_range = q3+(1.5*IQR)
```

```python
min_range,max_range
```

```python
dataset = dataset[dataset["CoapplicantIncome"]<=max_range]
```

```python
from sklearn.preprocessing import FunctionTransformer
```

```python
ft = FunctionTransformer(func= np.log1p)
```

```python
ft.fit(dataset[["CoapplicantIncome"]])
```

```python
dataset["CoapplicantIncome_tf"]=ft.fit_transform(dataset[["CoapplicantIncome"]])
```

```python
plt.subplot(1,2,1)
sns.histplot(dataset["CoapplicantIncome"])
plt.subplot(1,2,2)
sns.histplot(dataset["CoapplicantIncome_tf"])
plt.show()
```
