```python
import pandas as pd
```

```python
df = pd.DataFrame ({"size":["xl","m","l","xl","m","l","s","s","xl","xl","m","l"]})
```

```python
df.head(3)
```

```python
ord_data =[["s","m","l","xl"]]
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories=ord_data)
oe.fit_transform(df[["size"]])
```

```python
df["size_en"] = oe.fit_transform(df[["size"]])
```
